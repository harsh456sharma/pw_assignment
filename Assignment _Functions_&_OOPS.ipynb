```python
# 4. What are *args and **kwargs

# *args (Non-keyword variable arguments)
# Allows a function to accept any number of positional arguments.

# It collects extra arguments as a tuple.

def add_numbers(*args):
    total = sum(args)
    print("Sum:", total)

add_numbers(2, 3)
add_numbers(1, 2, 3, 4, 5)

#  **kwargs (Keyword variable arguments)
# Allows a function to accept any number of keyword arguments (i.e., key=value).

# It collects them as a dictionary.
def print_student_info(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

print_student_info(name="Harsh", age=22, course="B.Tech")

# Combined Use
# You can use both in the same function:
def show_info(*args, **kwargs):
    print("Positional:", args)
    print("Keyword:", kwargs)

show_info(10, 20, name="Harsh", city="Ranchi")
```

```python
# 5. Explain the iterator function

#  Iterator Function in Python – Explained
# An iterator in Python is an object that allows you to traverse through all the elements in a collection (like a list, tuple, or string), one item at a time.

# What Makes an Object an Iterator?
# An object is considered an iterator if it implements two methods:

# __iter__() – returns the iterator object itself.

# __next__() – returns the next item in the sequence. Raises StopIteration when items are exhausted

# A list is an iterable
my_list = [10, 20, 30]

# Convert to iterator using iter()
it = iter(my_list)

# Access elements using next()
print(next(it))  # Output: 10
print(next(it))  # Output: 20
print(next(it))  # Output: 30
# print(next(it))  # Raises StopIteration

 # Custom Iterator Example (Using Class):
class CountDown:
    def __init__(self, start):
        self.num = start

    def __iter__(self):
        return self

    def __next__(self):
        if self.num <= 0:
            raise StopIteration
        current = self.num
        self.num -= 1
        return current

cd = CountDown(3)

for number in cd:
    print(number)
```

```python
 # 6.Write a code that generates the squares of numbers from 1 to n using a generator.

# Generator function to yield squares from 1 to n
def generate_squares(n):
    for i in range(1, n + 1):
        yield i ** 2

# Example usage
n = 5
squares = generate_squares(n)

# Print the squares using a for loop
for square in squares:
    print(square)
```

```python
# 7. Write a code that generates palindromic numbers up to n using a generator
# Generator function to yield palindromic numbers up to n
def generate_palindromes(n):
    for i in range(1, n + 1):
        if str(i) == str(i)[::-1]:  # Check if number is a palindrome
            yield i

# Example usage
n = 200
for num in generate_palindromes(n):
    print(num)
```

```python
# 10.  Write a code that generates prime numbers up to n using a generator
# Helper function to check if a number is prime
def is_prime(num):
    if num <= 1:
        return False
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            return False
    return True

# Generator function to yield prime numbers up to n
def generate_primes(n):
    for i in range(2, n + 1):
        if is_prime(i):
            yield i

# Example usage
n = 30
for prime in generate_primes(n):
    print(prime)
```

```python
# 27. What is encapsulation in OOP

# Encapsulation is one of the fundamental concepts of Object-Oriented Programming (OOP). It refers to the bundling of data (attributes) and the methods (functions) that operate on that data into a single unit, called a class. Encapsulation also involves restricting direct access to some of an object's components, which can help prevent unintended interference and misuse of the data.

# Key Aspects of Encapsulation:
# Data Hiding (Access Control):

# In encapsulation, attributes (data) of an object are often made private (or protected), which means they cannot be accessed directly from outside the class. Instead, they are accessed through getter and setter methods (also called accessor and mutator methods).

# This allows for better control over how the data is modified, ensuring that invalid or unintended modifications do not occur.

# Public Interface:

# A class provides a public interface through which its functionality is accessed. This interface typically consists of methods that allow controlled access to the data within the class, hiding the internal implementation details.

# This provides a clean and simplified interface to the user of the class while keeping the internal logic safe from direct changes.


class BankAccount:
    def __init__(self, balance=0):
        self.__balance = balance  # Private attribute (Encapsulation)

    # Getter method
    def get_balance(self):
        return self.__balance

    # Setter method
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
        else:
            print("Deposit amount must be positive.")

    # Setter method
    def withdraw(self, amount):
        if amount > 0 and amount <= self.__balance:
            self.__balance -= amount
        else:
            print("Invalid withdrawal amount.")

# Create an instance of BankAccount
account = BankAccount(1000)

# Access the balance using the getter method
print("Initial Balance:", account.get_balance())

# Deposit and withdraw using setter methods
account.deposit(500)
account.withdraw(200)

# Access the updated balance using the getter method
print("Updated Balance:", account.get_balance())

# Attempting to directly access the private attribute will raise an error
# print(account.__balance)  # This will result in an AttributeError
```

```python
# 28. Explain the use of access modifiers in Python classes

# In Python, access modifiers are used to control the visibility or accessibility of attributes and methods within a class. Python does not have strict access control like other languages such as Java or C++, but it uses conventions to indicate the intended level of access. The access modifiers in Python are:

# 1. Public Access (no prefix):
# Public attributes and methods can be accessed freely from outside the class.

# By default, if no access modifier is used, the attribute or method is considered public.

class Person:
    def __init__(self, name, age):
        self.name = name  # Public attribute
        self.age = age    # Public attribute

    def greet(self):
        print(f"Hello, my name is {self.name} and I am {self.age} years old.")

# Accessing public attribute and method
person = Person("John", 30)
print(person.name)  # Access public attribute
person.greet()      # Call public method

# 2. Protected Access (_prefix):
# Attributes and methods that start with a single underscore (_) are considered protected by convention. This means they are intended for internal use within the class and its subclasses. While they can still be accessed directly, it is considered a bad practice to do so.

# Protected members are not strictly private, but they are meant to indicate that they should not be accessed from outside the class.

class Animal:
    def __init__(self, name):
        self._name = name  # Protected attribute

    def _speak(self):
        print(f"{self._name} makes a sound.")  # Protected method

class Dog(Animal):
    def speak(self):
        self._speak()  # Access protected method in subclass

dog = Dog("Buddy")
dog.speak()  # Calls the protected method from the subclass

# 3. Private Access (__prefix):
# Attributes and methods that start with a double underscore (__) are considered private. These members are not accessible directly from outside the class.

# Python uses a technique called name mangling to change the name of private attributes/methods, making them more difficult to access directly.

# Private members are meant to be used only within the class. However, it's important to note that Python does not enforce strict access control like other languages, so private members can still be accessed, but it is not recommended.

class Car:
    def __init__(self, model, year):
        self.__model = model  # Private attribute
        self.__year = year    # Private attribute

    def __private_method(self):
        print(f"Private method called for {self.__model}.")

    def display_info(self):
        print(f"Model: {self.__model}, Year: {self.__year}")

car = Car("Tesla", 2023)
car.display_info()

# Trying to access private attributes and methods directly will raise an error
# print(car.__model)  # AttributeError: 'Car' object has no attribute '__model'
# car.__private_method()  # AttributeError: 'Car' object has no attribute '__private_method'

```

```python
 # 29. What is inheritance in OOP?


# Inheritance is one of the core principles of Object-Oriented Programming (OOP). It allows a new class (called a child class or subclass) to inherit attributes and methods from an existing class (called a parent class or superclass). Inheritance promotes code reusability and the creation of a hierarchical class structure, making it easier to maintain and extend code.

# Key Concepts of Inheritance:
# Parent Class (Base Class):

# The class whose properties and behaviors (methods) are inherited by another class.

# Child Class (Derived Class):

# The class that inherits properties and behaviors from the parent class and can also have additional properties or methods or override the inherited ones.

# Method Overriding:

# A child class can modify or override methods inherited from the parent class to provide a different implementation.

# Super():

# The super() function is used to call methods from the parent class. It helps to access the parent class’s methods and constructors.

# Types of Inheritance:
# Single Inheritance: A class inherits from one parent class.

# Multiple Inheritance: A class inherits from more than one parent class.

# Multilevel Inheritance: A class inherits from a class that is also a child of another class.

# Hierarchical Inheritance: Multiple classes inherit from a single parent class.

# Hybrid Inheritance: A combination of two or more types of inheritance.

# Example of Inheritance in Python (Single Inheritance):

# Parent class (Base class)
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        print(f"{self.name} makes a sound.")

# Child class (Derived class)
class Dog(Animal):
    def __init__(self, name, breed):
        # Call the parent class constructor (super)
        super().__init__(name)
        self.breed = breed

    # Override the speak method
    def speak(self):
        print(f"{self.name} barks.")

# Create an instance of the Dog class
dog = Dog("Buddy", "Golden Retriever")
dog.speak()  # Call overridden method from Dog class

# Accessing the inherited name attribute
print(f"{dog.name} is a {dog.breed}.")


# Explanation:
# Animal is the parent class that has a method speak() and an attribute name.

# Dog is the child class that inherits from Animal:

# It inherits the name attribute and the speak() method from the Animal class.

# The speak() method is overridden in the Dog class to provide a specific implementation for dogs (i.e., barking).

# The __init__() method in Dog calls the parent class’s constructor using super() to initialize the name attribute.

# Example of Multiple Inheritance:

# Parent class 1 (Base class)
class Animal:
    def __init__(self, name):
        self.name = name

# Parent class 2 (Base class)
class Pet:
    def __init__(self, owner):
        self.owner = owner

# Child class (Derived class)
class Dog(Animal, Pet):
    def __init__(self, name, owner, breed):
        Animal.__init__(self, name)  # Initialize Animal class
        Pet.__init__(self, owner)    # Initialize Pet class
        self.breed = breed

    def speak(self):
        print(f"{self.name} barks!")

# Create an instance of the Dog class
dog = Dog("Buddy", "John", "Golden Retriever")
dog.speak()

# Access inherited attributes
print(f"{dog.name} belongs to {dog.owner} and is a {dog.breed}.")

# Explanation:
# Animal and Pet are two parent classes.

# Dog inherits from both Animal and Pet, so it has access to both the name (from Animal) and owner (from Pet) attributes.

# The speak() method is provided in the Dog class to define specific behavior for dogs.

# Benefits of Inheritance:
# Code Reusability: By inheriting from a parent class, a child class can reuse the code from the parent class without duplicating it.

# Extensibility: Child classes can add new features or modify existing ones, providing flexibility and extensibility.

# Maintenance: Changes made to the parent class automatically propagate to all child classes, making maintenance easier.

# Hierarchical Structure: It creates a clear hierarchical structure for classes, making code more organized and easier to understand.

# Method Resolution Order (MRO):
# In cases of multiple inheritance, Python uses a Method Resolution Order (MRO) to determine which method to call if there is ambiguity. It follows a left-to-right order in the class hierarchy to resolve method calls.

# Example of Method Resolution Order (MRO):

class A:
    def greet(self):
        print("Hello from class A.")

class B(A):
    def greet(self):
        print("Hello from class B.")

class C(A):
    def greet(self):
        print("Hello from class C.")

class D(B, C):
    pass

d = D()
d.greet()  # Which greet method will be called?

```

```python
# 30. Define polymorphism in OOP.

# Polymorphism is one of the key principles of Object-Oriented Programming (OOP). The term "polymorphism" comes from the Greek words "poly" (meaning "many") and "morph" (meaning "form"). In the context of OOP, polymorphism allows objects of different classes to be treated as objects of a common base class, while still allowing each subclass to have its own unique behavior.

# In simpler terms, polymorphism means that different objects can respond to the same method call in different ways. It is the ability of a single function, method, or operator to work with objects of different types.

# Key Types of Polymorphism in OOP:
# Method Overloading (Compile-time Polymorphism):

# In some languages, method overloading allows you to define multiple methods with the same name but different parameters (number or type). However, Python does not support method overloading directly, as it allows default arguments or variable arguments to achieve similar functionality.

# Method Overriding (Runtime Polymorphism):

# Method overriding occurs when a subclass provides its own specific implementation of a method that is already defined in its parent class. The method in the child class overrides the method in the parent class, and the version of the method that is executed depends on the object's actual class.

# Operator Overloading (Custom Operators):

# In some cases, polymorphism can be used to define custom behavior for operators (like +, -, *, etc.) when applied to objects of a class.

# Polymorphism in Python:
# In Python, polymorphism is primarily implemented through method overriding and duck typing.

# Duck Typing: "If it looks like a duck, swims like a duck, and quacks like a duck, it must be a duck." In Python, you don't need to explicitly declare that a class implements an interface. As long as an object implements a method or behavior, it can be treated as a type that supports that method.

# Parent Class (Base class)
class Animal:
    def speak(self):
        print("The animal makes a sound.")

# Child Class (Derived class)
class Dog(Animal):
    def speak(self):
        print("The dog barks.")

# Another Child Class (Derived class)
class Cat(Animal):
    def speak(self):
        print("The cat meows.")

# Create instances of Dog and Cat
animal = Animal()
dog = Dog()
cat = Cat()

# Polymorphism in action: same method name, different behaviors
animal.speak()  # Calls speak() from Animal class
dog.speak()     # Calls speak() from Dog class
cat.speak()     # Calls speak() from Cat class

# Polymorphism Through Duck Typing:
# In Python, polymorphism can also be achieved through duck typing. This allows objects to behave like other objects as long as they implement the required methods.

class Bird:
    def fly(self):
        print("The bird flies.")

class Airplane:
    def fly(self):
        print("The airplane flies.")

# Function that expects an object with a fly method
def make_it_fly(flyable_object):
    flyable_object.fly()

# Create instances of Bird and Airplane
bird = Bird()
airplane = Airplane()

# Polymorphism in action: Both objects have the same method, so they can be treated the same
make_it_fly(bird)     # Outputs: The bird flies.
make_it_fly(airplane) # Outputs: The airplane flies.

```

```python
# 31. Explain method overriding in Python

# Method Overriding in Python is a concept where a subclass provides a specific implementation of a method that is already defined in its superclass (parent class). The subclass's method overrides the method in the parent class, allowing it to change or extend the behavior of that method.

# Key Points:
# Inheritance: Method overriding occurs in an inheritance hierarchy, where a subclass inherits from a parent class but needs to change or enhance a method’s behavior.

# Same Method Signature: The method in the subclass must have the same name, same parameters, and same return type as the method in the parent class.

# Runtime Polymorphism: The version of the method that gets called is determined at runtime based on the actual object (not the reference type), which is a feature of polymorphism.

# Why Use Method Overriding?
# Customization: You can customize or change the behavior of methods that are inherited from a parent class.

# Specialization: It allows subclasses to specialize or adapt general behavior from the parent class.

# Flexibility: It provides the flexibility to define behavior in a more specific context while keeping the parent class structure intact.

# Parent class (Base class)
class Animal:
    def speak(self):
        print("The animal makes a sound.")

# Child class (Derived class)
class Dog(Animal):
    def speak(self):
        print("The dog barks.")

# Another child class (Derived class)
class Cat(Animal):
    def speak(self):
        print("The cat meows.")

# Creating instances of Animal, Dog, and Cat
animal = Animal()
dog = Dog()
cat = Cat()

# Calling the speak() method on each instance
animal.speak()  # Output: The animal makes a sound.
dog.speak()     # Output: The dog barks.
cat.speak()     # Output: The cat meows.

# Using super() in Method Overriding:
# In some cases, you may want to call the method from the parent class in addition to the overridden behavior in the subclass. This can be done using the super() function, which allows access to the parent class's methods.

# Parent class (Base class)
class Animal:
    def speak(self):
        print("The animal makes a sound.")

# Child class (Derived class)
class Dog(Animal):
    def speak(self):
        super().speak()  # Calling the parent class's speak method
        print("The dog barks.")

# Creating an instance of Dog
dog = Dog()
dog.speak()  # Output: The animal makes a sound. \n The dog barks.


# Overriding Constructor (__init__):
# In Python, you can also override the constructor (__init__) of the parent class in a subclass. This allows you to initialize or customize attributes when creating an object of the subclass.

# Parent class (Base class)
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        print(f"{self.name} makes a sound.")

# Child class (Derived class)
class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name)  # Calling the parent class's constructor
        self.breed = breed

    def speak(self):
        print(f"{self.name} barks.")

# Creating an instance of Dog
dog = Dog("Buddy", "Golden Retriever")
dog.speak()  # Output: Buddy barks.
print(f"{dog.name} is a {dog.breed}.")  # Output: Buddy is a Golden Retriever.

```

```python
# 37. What is abstraction in Python? How is it implemented?

# Abstraction in Python
# Abstraction is one of the fundamental concepts in Object-Oriented Programming (OOP) that allows us to hide the complex implementation details and show only the essential features of an object. In Python, abstraction helps to reduce complexity by providing a clear interface to interact with the object, while hiding the underlying logic.

# Key Points:
# Abstraction focuses on what an object does, rather than how it does it.

# It allows us to define a clear interface for the user without exposing the internal workings.

# Abstraction is implemented using abstract classes and abstract methods.

# How is Abstraction Implemented in Python?
# Python supports abstraction through the use of abstract classes and abstract methods, which are defined in the abc (Abstract Base Class) module.

# Abstract Class: A class that cannot be instantiated and is meant to be subclassed. It may contain abstract methods that must be implemented by subclasses.

# Abstract Method: A method declared in an abstract class, which does not have an implementation. Subclasses must provide an implementation for these methods.

# Example of Abstraction in Python:

from abc import ABC, abstractmethod

# Abstract class Animal
class Animal(ABC):
    
    @abstractmethod
    def sound(self):
        pass  # No implementation, just an abstract method

    @abstractmethod
    def move(self):
        pass  # No implementation, just an abstract method

# Subclass Dog implements abstract methods from Animal
class Dog(Animal):
    
    def sound(self):
        print("Woof!")
        
    def move(self):
        print("Dog runs")

# Subclass Cat implements abstract methods from Animal
class Cat(Animal):
    
    def sound(self):
        print("Meow!")
        
    def move(self):
        print("Cat jumps")

# Now, we can instantiate objects of Dog and Cat
dog = Dog()
cat = Cat()

# Call methods
dog.sound()  # Output: Woof!
dog.move()   # Output: Dog runs

cat.sound()  # Output: Meow!
cat.move()   # Output: Cat jumps
```

```python
# 42. How does Python achieve polymorphism through method overriding

# Python achieves polymorphism through method overriding by allowing a child (derived) class to provide a specific implementation of a method that is already defined in its parent (base) class. This allows objects of different classes to be treated through a common interface while behaving differently based on their actual type.

# 💡 What is Polymorphism?
# Polymorphism means "many forms." In Object-Oriented Programming (OOP), it refers to the ability to call the same method on different objects, and each object responds in its own way depending on its class.

# 🔁 How Method Overriding Enables Polymorphism in Python:
# A base class defines a method (e.g., speak()).

# A derived class (e.g., Dog, Cat) overrides that method with its own implementation.

# When you call the method on a derived class object via a reference to the base class, the correct version is invoked (this is runtime polymorphism).

# ✅ Example of Method Overriding and Polymorphism in Python:

# Base class
class Animal:
    def speak(self):
        print("Animal makes a sound")

# Derived class 1
class Dog(Animal):
    def speak(self):
        print("Dog barks")

# Derived class 2
class Cat(Animal):
    def speak(self):
        print("Cat meows")

# Function that demonstrates polymorphism
def make_animal_speak(animal: Animal):
    animal.speak()

# Creating objects
dog = Dog()
cat = Cat()

# Using the same function to invoke overridden methods
make_animal_speak(dog)  # Output: Dog barks
make_animal_speak(cat)  # Output: Cat meows
```

```python
# 47. How do you achieve encapsulation in Python?

# Encapsulation in Python is achieved by restricting direct access to some of an object’s internal components and bundling data (attributes) and methods (functions) that operate on that data into a single unit — typically a class.

# Ways to Achieve Encapsulation in Python:
# 1. Using Underscores for Attribute Privacy
# Syntax	Access Level	Purpose
# self.name	Public	Can be accessed from anywhere
# self._name	Protected (by convention)	Should not be accessed directly outside the class
# self.__name	Private (name mangling)	Stronger restriction against external access

class Person:
    def __init__(self, name, age):
        self.name = name        # Public
        self._age = age         # Protected (convention)
        self.__ssn = "123-45-6789"  # Private (name mangling)

    def get_ssn(self):
        return self.__ssn

    def set_ssn(self, new_ssn):
        if isinstance(new_ssn, str) and len(new_ssn) == 11:
            self.__ssn = new_ssn
        else:
            print("Invalid SSN format")

p = Person("Alice", 30)

print(p.name)        # Allowed
print(p._age)        # Allowed but discouraged
# print(p.__ssn)     # ❌ AttributeError
print(p.get_ssn())   # ✅ Use public method to access private data
```

```python
# 48. Can encapsulation be bypassed in Python? If so, how?

# 1. Accessing Protected Members (_member)
# In Python, a single underscore (_var) signals that the variable is intended to be protected.

# But this is only a convention — you can still access it from outside the class.

class Person:
    def __init__(self):
        self._age = 25  # Protected by convention

p = Person()
print(p._age)  # ❗ Can be accessed, but discouraged


# 2. Accessing Private Members (__member) via Name Mangling
# Python “hides” private members by name mangling: it renames __var to _ClassName__var.

# But you can still access them if you know the mangled name.

class Person:
    def __init__(self):
        self.__ssn = "123-45-6789"  # Private

p = Person()
print(p._Person__ssn)  # ❗ Bypasses encapsulation using name mangling
```

```python
# 49.  Implement a class BankAccount with a private balance attribute. Include methods to deposit, withdraw,
# and check the balance.

class BankAccount:
    def __init__(self, initial_balance=0):
        self.__balance = initial_balance  # Private attribute

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited: ${amount}")
        else:
            print("Deposit amount must be positive.")

    def withdraw(self, amount):
        if amount > 0:
            if amount <= self.__balance:
                self.__balance -= amount
                print(f"Withdrew: ${amount}")
            else:
                print("Insufficient balance.")
        else:
            print("Withdrawal amount must be positive.")

    def check_balance(self):
        return self.__balance


account = BankAccount(100)  # Initial balance = $100
account.deposit(50)         # Deposit $50
account.withdraw(30)        # Withdraw $30
print("Current balance:", account.check_balance())  # Output: 120

```

```python
# 66. Explain the usage of the __iter__() and __next__() methods in Python. Provide an example using
# iterators.

# In Python, the __iter__() and __next__() methods are part of the iterator protocol, which allows objects to be iterated over in a loop (e.g., in a for loop). These methods are used for creating custom iterable objects, enabling you to define how your objects should behave when iterated over, such as in loops or with functions like next().

# 1. __iter__() Method:
# The __iter__() method is responsible for returning an iterator object from an iterable object. It is called when an object is passed to an iterator (e.g., a for loop). This method must return an iterator, which is an object that implements the __next__() method.

# 2. __next__() Method:
# The __next__() method is used to return the next item in the sequence. When the next() function is called on an iterator, it invokes the __next__() method. If there are no more items to return, __next__() should raise the StopIteration exception, signaling the end of the iteration.

# Usage:
# __iter__() is called to initialize the iteration process, returning the iterator.

# __next__() is called to fetch the next item during the iteration process.

class Countdown:
    def __init__(self, start):
        self.start = start  # The starting point of the countdown
    
    def __iter__(self):
        self.current = self.start  # Initialize the iteration state
        return self  # Return the iterator object itself
    
    def __next__(self):
        if self.current <= 0:
            raise StopIteration  # Stop iteration when current reaches 0
        self.current -= 1  # Decrement the counter
        return self.current  # Return the current value

# Create an instance of Countdown
countdown = Countdown(5)

# Iterate over the Countdown object
for number in countdown:
    print(number)


#  Explanation:
# __init__(): Initializes the countdown starting number.

# __iter__(): This method is called when the object is used in an iteration (like a for loop). It initializes the current counter and returns the object itself, as it is the iterator.

# __next__(): Each time __next__() is called (during iteration), it decrements the current value and returns it. If the countdown reaches 0, it raises the StopIteration exception, signaling the end of the iteration.

# How It Works:
# When the for loop is executed, Python calls the __iter__() method of the Countdown class to get an iterator (which in this case is the object itself).

# Then, the for loop repeatedly calls the __next__() method to get the next value.

# When the StopIteration exception is raised (after the countdown reaches 0), the loop ends.

# Example with Custom Range:
# Here’s another example using __iter__() and __next__() to implement a custom range function:

class MyRange:
    def __init__(self, start, end):
        self.start = start
        self.end = end
    
    def __iter__(self):
        self.current = self.start  # Initialize current to start
        return self
    
    def __next__(self):
        if self.current >= self.end:
            raise StopIteration  # End iteration when current reaches end
        self.current += 1  # Increment the counter
        return self.current - 1  # Return the current value before incrementing

# Create an instance of MyRange
my_range = MyRange(1, 5)

# Iterate over the custom range object
for number in my_range:
    print(number)

# Explanation:
# __init__(): Initializes the range with a start and end value.

# __iter__(): Prepares the iterator (sets current to start).

# __next__(): Returns the next value in the range and increments the current value. If current reaches end, it raises StopIteration.
```

```python
# 67. What is the purpose of a getter method in Python? Provide an example demonstrating the use of a getter
# method using property decorators

# In Python, a getter method is used to access the value of a private or protected attribute from outside the class. This is part of the encapsulation principle in object-oriented programming, where you may want to restrict direct access to an attribute but still provide a way to retrieve its value.

# Purpose of Getter Methods:
# Encapsulation: Getter methods are used to control access to private attributes, ensuring that their values are not changed directly or improperly.

# Read-Only Access: You can use a getter to provide read-only access to an attribute, ensuring that users can access the value but cannot modify it directly.

# Validation: A getter can include logic to modify or validate the value before returning it. For example, you might compute or format the attribute before returning it.

# Property Decorators: In Python, the property decorator is a more Pythonic way to create getter methods. This allows you to define methods that act like attributes, giving you the flexibility to modify their behavior without changing the interface.

class Circle:
    def __init__(self, radius):
        self._radius = radius  # private attribute

    @property
    def radius(self):
        """Getter method using property decorator"""
        return self._radius

    @property
    def area(self):
        """A computed property: Getter method for area"""
        return 3.14 * self._radius ** 2

    @property
    def diameter(self):
        """A computed property: Getter method for diameter"""
        return 2 * self._radius

# Create an instance of Circle
circle = Circle(5)

# Access the radius (using the getter method)
print(f"Radius: {circle.radius}")  # Output: Radius: 5

# Access the area (using the getter method)
print(f"Area: {circle.area}")  # Output: Area: 78.5

# Access the diameter (using the getter method)
print(f"Diameter: {circle.diameter}")  # Output: Diameter: 10


#  Explanation:
# __init__(self, radius): This is the constructor that initializes the Circle object with a radius value. The _radius attribute is considered "protected" by convention (using an underscore) to signal that it should not be accessed directly.

# @property def radius(self): This is the getter method for the radius attribute. Using the @property decorator, we can access it like an attribute (e.g., circle.radius) instead of a method (e.g., circle.radius()).

# @property def area(self): This is a computed property that calculates the area of the circle based on its radius. When accessed, it returns the computed value of the area, allowing us to retrieve it as if it were a regular attribute.

# @property def diameter(self): This is another computed property that calculates the diameter of the circle.

# Why Use Property Decorators?
# Convenience: Using the property decorator allows you to use methods as if they are attributes, which makes the code more readable and clean.

# Encapsulation: By using the property decorator, you encapsulate the logic for accessing and computing values, allowing you to change the underlying implementation without affecting how the class is used externally.

# No Direct Access: Users of the class can access circle.radius, circle.area, and circle.diameter just like regular attributes, but behind the scenes, the getter methods are invoked to retrieve the values.
```

```python
# 69. What is the purpose of the @property decorator in Python? Provide an example illustrating its usage.

# The @property decorator in Python is used to define a method as a property, allowing it to be accessed like an attribute rather than a method. This decorator provides a way to define getter methods that can be accessed like regular attributes while still giving you the flexibility to include additional logic for retrieving the value.

# Purpose of the @property Decorator:
# Encapsulation: It helps in hiding the internal implementation of an attribute while still providing access to it. It allows you to modify how the attribute is retrieved without changing the interface of the class.

# Cleaner Code: Using the @property decorator allows you to define getter methods that act like attributes, keeping the code more concise and readable.

# Validation and Computation: It enables you to perform validation or compute values dynamically before returning them, without exposing the internal logic to the user of the class.

# How It Works:
# When you use the @property decorator, it allows you to define a method that can be accessed just like an attribute.

# You can also define setter and deleter methods (if needed) using @property.setter and @property.deleter.

# Example of @property Decorator Usage:

class Circle:
    def __init__(self, radius):
        self._radius = radius  # private attribute

    @property
    def radius(self):
        """Getter method for radius"""
        return self._radius

    @radius.setter
    def radius(self, value):
        """Setter method for radius to ensure it's positive"""
        if value <= 0:
            raise ValueError("Radius must be positive")
        self._radius = value

    @property
    def area(self):
        """Computed property for area"""
        return 3.14 * self._radius ** 2

    @property
    def diameter(self):
        """Computed property for diameter"""
        return 2 * self._radius


# Create an instance of Circle
circle = Circle(5)

# Access the radius using the getter method
print(f"Radius: {circle.radius}")  # Output: Radius: 5

# Modify the radius using the setter method
circle.radius = 10
print(f"Modified Radius: {circle.radius}")  # Output: Modified Radius: 10

# Try setting a negative radius (will raise an exception)
try:
    circle.radius = -3
except ValueError as e:
    print(e)  # Output: Radius must be positive

# Access the area and diameter (computed properties)
print(f"Area: {circle.area}")  # Output: Area: 314.0
print(f"Diameter: {circle.diameter}")  # Output: Diameter: 20


#  Explanation:
# __init__(self, radius): This initializes the Circle object with a given radius, storing it in the private attribute _radius.

# @property def radius(self): This is the getter method for the radius attribute. When you access circle.radius, it invokes this method and returns the value of _radius. This makes it look like an attribute, but it's a method behind the scenes.

# @radius.setter def radius(self, value): This is the setter method for the radius attribute. It allows you to modify the radius but includes validation logic to ensure the radius is positive. If you try to set a negative radius, it raises a ValueError.

# @property def area(self): This is a computed property that calculates the area of the circle based on its radius. When you access circle.area, it calculates the area dynamically using the formula π * r^2.

# @property def diameter(self): This is another computed property that calculates the diameter of the circle.
```

```python
# 70. Explain the use of the @deleter decorator in Python property decorators. Provide a code example
# demonstrating its application.

# The @deleter decorator in Python is part of the property mechanism and is used to define a method for deleting an attribute. When you use the @deleter decorator, you specify what should happen when the del statement is called on an attribute. This allows you to customize how an attribute is deleted, for example, by raising an exception or performing some cleanup task.

# Like @property and @<property_name>.setter, the @deleter decorator is applied to a method that is linked to an attribute. This method is executed when you delete the attribute using the del keyword.

class Person:
    def __init__(self, name, age):
        self._name = name
        self._age = age

    # Getter for name
    @property
    def name(self):
        return self._name

    # Setter for name
    @name.setter
    def name(self, value):
        if len(value) < 3:
            raise ValueError("Name must be at least 3 characters long.")
        self._name = value

    # Getter for age
    @property
    def age(self):
        return self._age

    # Setter for age
    @age.setter
    def age(self, value):
        if value < 0:
            raise ValueError("Age must be a positive integer.")
        self._age = value

    # Deleter for name
    @name.deleter
    def name(self):
        print("Deleting the name attribute...")
        del self._name

    # Deleter for age
    @age.deleter
    def age(self):
        print("Deleting the age attribute...")
        del self._age


# Create an instance of Person
person = Person("John", 25)

# Accessing the attributes
print(person.name)  # Output: John
print(person.age)   # Output: 25

# Deleting the name attribute
del person.name  # Output: Deleting the name attribute...

# Deleting the age attribute
del person.age  # Output: Deleting the age attribute...
```

```python
# 71. How does encapsulation relate to property decorators in Python? Provide an example showcasing
# encapsulation using property decorators.

# Encapsulation and Property Decorators in Python
# Encapsulation is one of the core principles of object-oriented programming (OOP). It refers to restricting direct access to some of an object's components and only allowing controlled interaction through methods. This helps protect the internal state of an object and prevents external code from corrupting it.

# In Python, encapsulation is typically implemented using:

# Private attributes (by convention, prefixed with _ or __)

# Property decorators (@property, @<property>.setter, @<property>.deleter)

# Property decorators provide a clean and Pythonic way to encapsulate the internal state of a class while still exposing it like a normal attribute.


class BankAccount:
    def __init__(self, owner, balance):
        self._owner = owner
        self._balance = balance  # Encapsulated attribute

    # Getter method
    @property
    def balance(self):
        return self._balance

    # Setter method with validation
    @balance.setter
    def balance(self, amount):
        if amount < 0:
            raise ValueError("Balance cannot be negative.")
        self._balance = amount

    # Getter for owner
    @property
    def owner(self):
        return self._owner

    # No setter for owner – read-only

# Create an account
account = BankAccount("Alice", 1000)

# Access balance using getter
print(account.balance)  # Output: 1000

# Modify balance using setter
account.balance = 1500
print(account.balance)  # Output: 1500

# Try setting a negative balance (will raise error)
try:
    account.balance = -500
except ValueError as e:
    print(e)  # Output: Balance cannot be negative.

# Access owner (read-only)
print(account.owner)  # Output: Alice

# Attempting to modify owner will raise an AttributeError
try:
    account.owner = "Bob"
except AttributeError as e:
    print(e)  # Output: can't set attribute
```
