```python
# 4. Generate a dataset with categories and numerical values. Visualize the distribution of a numerical variable across different categories.

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Set random seed for reproducibility
np.random.seed(42)

# Generate synthetic dataset
categories = ['A', 'B', 'C', 'D']
data = []

for category in categories:
    # Generate 50 random values for each category
    values = np.random.normal(loc=np.random.randint(40, 60), scale=5, size=50)
    for value in values:
        data.append([category, value])

# Convert to DataFrame
df = pd.DataFrame(data, columns=['Category', 'Value'])

# Display first few rows (optional)
print(df.head())

# Visualize the distribution using a box plot
plt.figure(figsize=(8,6))
df.boxplot(column='Value', by='Category', grid=False)
plt.title('Distribution of Values Across Categories')
plt.suptitle('')  # Remove automatic suptitle
plt.xlabel('Category')
plt.ylabel('Value')
plt.show()

import seaborn as sns

plt.figure(figsize=(8,6))
sns.violinplot(x='Category', y='Value', data=df, palette='pastel')
plt.title('Violin Plot: Distribution of Values Across Categories')
plt.xlabel('Category')
plt.ylabel('Value')
plt.show()

```

```python
# 5. Generate a synthetic dataset with correlated features. Visualize the correlation matrix of a dataset using a heatmap.

import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Set random seed for reproducibility
np.random.seed(42)

# Generate synthetic correlated data
# Base feature
x = np.random.normal(0, 1, 500)

# Create features correlated with x
y = 2 * x + np.random.normal(0, 0.5, 500)
z = -1.5 * x + np.random.normal(0, 0.5, 500)
w = x + y + np.random.normal(0, 0.5, 500)

# Combine into a DataFrame
df = pd.DataFrame({
    'Feature_X': x,
    'Feature_Y': y,
    'Feature_Z': z,
    'Feature_W': w
})

# View first few rows (optional)
print(df.head())

# Compute correlation matrix
corr_matrix = df.corr()

# Visualize correlation matrix using a heatmap
plt.figure(figsize=(8,6))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f', square=True)
plt.title('Correlation Matrix Heatmap')
plt.show()


```

```python
# 3. Using the sales data, generate a heatmap to visualize the variation in sales across different months and days.

# np.random.seed(20)
# data = {
#     'Month': np.random.choice(['Jan', 'Feb', 'Mar', 'Apr', 'May'], 100),
#     'Day': np.random.choice(range(1, 31), 100),
#     'Sales': np.random.randint(1000, 5000, 100)
# }
# df = pd.DataFrame(data)


import numpy as np
import pandas as pd

# Generate sales data
np.random.seed(20)
data = {
    'Month': np.random.choice(['Jan', 'Feb', 'Mar', 'Apr', 'May'], 100),
    'Day': np.random.choice(range(1, 31), 100),
    'Sales': np.random.randint(1000, 5000, 100)
}

df = pd.DataFrame(data)

# Create a pivot table: rows = Days, columns = Months
pivot_table = df.pivot_table(index='Day',
                             columns='Month',
                             values='Sales',
                             aggfunc='mean').fillna(0)

# Sort days for better heatmap layout
pivot_table = pivot_table.sort_index()

import plotly.express as px

fig = px.imshow(
    pivot_table,
    labels=dict(x="Month", y="Day", color="Average Sales"),
    x=pivot_table.columns,
    y=pivot_table.index,
    text_auto=True,
    color_continuous_scale='Viridis',
    title='Heatmap: Variation in Sales Across Months and Days'
)

fig.update_layout(
    xaxis_title="Month",
    yaxis_title="Day"
)

fig.show()


```

```python
# 4. Using the given x and y data, generate a 3D surface plot to visualize the function z = sin(sqrt{x^2 + y^2})

# x = np.linspace(-5, 5, 100)
# y = np.linspace(-5, 5, 100)
# x, y = np.meshgrid(x, y)
# z = np.sin(np.sqrt(x**2 + y**2))
# data = {
#     'X': x.flatten(),
#     'Y': y.flatten(),
#     'Z': z.flatten()
# }
# df = pd.DataFrame(data)

import numpy as np
import pandas as pd
import plotly.graph_objects as go

# ------------------------------
# Generate the meshgrid and compute Z
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
x, y = np.meshgrid(x, y)
z = np.sin(np.sqrt(x**2 + y**2))

# ------------------------------
# Flatten for a DataFrame (optional)
data = {
    'X': x.flatten(),
    'Y': y.flatten(),
    'Z': z.flatten()
}
df = pd.DataFrame(data)

# ------------------------------
# Create the 3D surface plot using go.Surface
fig = go.Figure(data=[go.Surface(z=z, x=x, y=y, colorscale='Viridis')])

# Add layout details
fig.update_layout(
    title='3D Surface Plot: z = sin(sqrt(x^2 + y^2))',
    scene=dict(
        xaxis_title='X',
        yaxis_title='Y',
        zaxis_title='Z'
    )
)

fig.show()
```

```python
# 1.Create a Bokeh plot displaying a sine wave. Set x-values from 0 to 10 and y-values as the sine of x.

import numpy as np
from bokeh.plotting import figure, show
from bokeh.io import output_notebook

# If you’re using a notebook, display inline
output_notebook()

# ------------------------------
# Generate data
x = np.linspace(0, 10, 500)
y = np.sin(x)

# ------------------------------
# Create a Bokeh figure
p = figure(title="Sine Wave",
           x_axis_label='X',
           y_axis_label='sin(X)',
           width=700, height=400)

# Add a line glyph
p.line(x, y, legend_label="sin(x)", line_width=2, line_color="navy")

# Show the plot
show(p)
```

```python
# 2. Create a Bokeh scatter plot using randomly generated x and y values. Use different sizes and colors for the markers based on the 'sizes' and 'colors' columns.

import numpy as np
import pandas as pd
from bokeh.plotting import figure, show
from bokeh.io import output_notebook
from bokeh.models import ColumnDataSource
from bokeh.palettes import Category10

# Display plots inline in a notebook
output_notebook()

# ------------------------------
# Generate random data
np.random.seed(42)
N = 100

df = pd.DataFrame({
    'x': np.random.uniform(0, 100, N),
    'y': np.random.uniform(0, 100, N),
    'sizes': np.random.uniform(5, 30, N),
    'colors': np.random.choice(Category10[10], N)
})

# ------------------------------
# Create a ColumnDataSource
source = ColumnDataSource(df)

# ------------------------------
# Create a Bokeh scatter plot
p = figure(title="Random Scatter Plot with Sizes and Colors",
           x_axis_label='X',
           y_axis_label='Y',
           width=700, height=500)

p.scatter(x='x', y='y',
          size='sizes',
          color='colors',
          fill_alpha=0.6,
          source=source,
          legend_label='Points')

# Optional: Adjust legend
p.legend.location = 'top_left'

# Show the plot
show(p)
```

```python
#  3. Generate a Bokeh bar chart representing the counts of different fruits using the following dataset.

# fruits = ['Apples', 'Oranges', 'Bananas', 'Pears']
# counts = [20, 25, 30, 35]

from bokeh.plotting import figure, show
from bokeh.io import output_notebook
from bokeh.transform import dodge
from bokeh.models import ColumnDataSource

# Display plot inline (for Jupyter)
output_notebook()

# ------------------------------
# Your data
fruits = ['Apples', 'Oranges', 'Bananas', 'Pears']
counts = [20, 25, 30, 35]

source = ColumnDataSource(data=dict(fruits=fruits, counts=counts))

# ------------------------------
# Create figure
p = figure(x_range=fruits,
           height=400,
           width=600,
           title="Fruit Counts",
           toolbar_location=None,
           tools="")

# Draw bars
p.vbar(x='fruits',
       top='counts',
       width=0.5,
       source=source,
       fill_color='skyblue',
       line_color='black')

# ------------------------------
# Styling
p.xgrid.grid_line_color = None
p.y_range.start = 0
p.yaxis.axis_label = "Count"
p.xaxis.axis_label = "Fruit"

# Show plot
show(p)
```

```python
# 4. Create a Bokeh histogram to visualize the distribution of the given data.

# data_hist = np.random.randn(1000)
# hist, edges = np.histogram(data_hist, bins=30)

import numpy as np
from bokeh.plotting import figure, show
from bokeh.io import output_notebook

# Display plot inline (for Jupyter)
output_notebook()

# ------------------------------
# Generate data and histogram
data_hist = np.random.randn(1000)
hist, edges = np.histogram(data_hist, bins=30)

# ------------------------------
# Create Bokeh figure
p = figure(title='Histogram of Data',
           x_axis_label='Value',
           y_axis_label='Frequency',
           width=700,
           height=400)

# Draw histogram using quad glyphs
p.quad(top=hist,
       bottom=0,
       left=edges[:-1],
       right=edges[1:],
       fill_color='navy',
       line_color='white',
       alpha=0.7)

# ------------------------------
# Show plot
show(p)
```

```python
# 5. Create a Bokeh heatmap using the provided dataset.

# data_heatmap = np.random.rand(10, 10)
# x = np.linspace(0, 1, 10)
# y = np.linspace(0, 1, 10)
# xx, yy = np.meshgrid(x, y)

import numpy as np
from bokeh.plotting import figure, show
from bokeh.io import output_notebook
from bokeh.models import ColumnDataSource, LinearColorMapper, ColorBar
from bokeh.transform import transform
from bokeh.palettes import Viridis256

# Display inline in notebook
output_notebook()

# ------------------------------
# Given data
np.random.seed(42)
data_heatmap = np.random.rand(10, 10)
x = np.linspace(0, 1, 10)
y = np.linspace(0, 1, 10)
xx, yy = np.meshgrid(x, y)

# Flatten data for Bokeh ColumnDataSource
source = ColumnDataSource(data=dict(
    x=xx.flatten(),
    y=yy.flatten(),
    value=data_heatmap.flatten()
))

# ------------------------------
# Define color mapper
color_mapper = LinearColorMapper(palette=Viridis256, low=data_heatmap.min(), high=data_heatmap.max())

# ------------------------------
# Create figure
p = figure(title="Heatmap Example",
           x_axis_label='X',
           y_axis_label='Y',
           x_range=(0, 1),
           y_range=(0, 1),
           width=500,
           height=500,
           tools='')

# Draw rectangles (heatmap cells)
p.rect(x='x', y='y',
       width=0.1, height=0.1,
       source=source,
       fill_color=transform('value', color_mapper),
       line_color=None)

# Add color bar
color_bar = ColorBar(color_mapper=color_mapper, location=(0, 0))
p.add_layout(color_bar, 'right')

# ------------------------------
# Show plot
show(p)
```
